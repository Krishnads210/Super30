```python
Products = ["Laptop", "Phone", "Mouse", "Keyboard", "Monitor", "Speaker"]

Prices  = {"Laptop" : 65000 , "Phone" : 25000, "Mouse" : 800, "Keyboard" : 1200, "Monitor" : 9000, "Speaker" : 2000}


Quantity = {}

Invoice = 0


for i in range(len(Products)):
  Choose_a_Product = input("Choose a Product(Laptop, Phone,Mouse, Keyboard, Monitor or Speaker):")

  if Choose_a_Product not in Products:
    print("Invalid Product")
    break
  elif Choose_a_Product in Quantity:
    print("Product already Exists")
    Do_you_want_to_update_quantity = input("Do you wan to update the quantity (Yes/No")
    if Do_you_want_to_update_quantity == "Yes":
      Enter_quantity =  int(input("Please choose quantity:"))
      Quantity[Choose_a_Product] = Enter_quantity
  else:
    Enter_quantity =  int(input("Please choose quantity:"))
    Quantity[Choose_a_Product] = Enter_quantity
  
  Choose_another_product = input("Choose another Product(Yes/No):")

  if Choose_another_product == "No":
    break

print(Quantity)

for i, v in Prices.items():
  for j, w in Quantity.items():
    if i == j:
      Invoice = Invoice + v * w
if Invoice > 60000:
  Final_Invoice = Invoice * 0.85
elif Invoice > 30000:
  Final_Invoice = Invoice * 0.90
elif Invoice > 10000:
  Final_Invoice = Invoice * 0.95
else:
  Final_Invoice = Invoice



print(Final_Invoice)
      
```
