```python
student = {}

n = input("Please enter the number of students:")

for i in range(1,int(n)+1):
    student[str(i)] = {}

for i in student.values():
    i["Name"] = ''
    i["Marks"] = []
    # Enter the name of the student
    Name = input("Please enter the name of student")
    # Enter the marks in 5 subjects
    Maths_Marks = int(input("Please enter the marks of the student in Maths:"))
    Science_Marks = int(input("Please enter the marks of the student in Science:"))
    English_Marks = int(input("Please enter the marks of the student in English:"))
    History_Marks = int(input("Please enter the marks of the student in History:"))
    Civics_Marks = int(input("Please enter the marks of the student in Civics:"))
    # Add the details to the student dictionary
    i["Name"] = Name
    i["Marks"].append(Maths_Marks)
    i["Marks"].append(Science_Marks)
    i["Marks"].append(English_Marks)
    i["Marks"].append(History_Marks)
    i["Marks"].append(Civics_Marks)

for i in student.values():
    i["Total_Marks"] = sum(i["Marks"])
    i["Percentage"] = round((sum(i["Marks"])/500)*100,2)
    if i["Percentage"] >= 90:
        i["Grade"] = "A"
    elif i["Percentage"] >= 80:
         i["Grade"] = "B"
    elif i["Percentage"] >= 70:
         i["Grade"] = "C"
    elif i["Percentage"] >= 60:
         i["Grade"] = "D"
    else:
         i["Grade"] = "F"

all_the_totals = []
all_the_percentages = []
failed_students = 0

for i in student.values():
  all_the_totals.append(i['Total_Marks'])
  all_the_percentages.append(i['Percentage'])
  if i['Grade'] == 'F':
    failed_students +=1

class_pct_avg = round(sum(all_the_percentages)/int(n),2)

Class_report = {'Highest_marks_student':'','Lowest_marks_student':'', 'Avg_pctage_class':class_pct_avg, 'failed_students': failed_students}

Highest_marks_student = ''
Lowest_marks_student = []
for i in student.values():
    if i['Total_Marks'] == max(all_the_totals):
        Class_report['Highest_marks_student'] = i['Name']
    if i['Total_Marks'] == min(all_the_totals):
        Class_report['Lowest_marks_student'] = i['Name']

# Below are the student details
print(student)

# Below is the class report
print(Class_report)
```
