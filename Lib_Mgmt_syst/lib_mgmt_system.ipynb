```python
# Structure: {book_name: {"author": "...", "available": True/False}}
library = {
    "To kill a mockingbird": {"author": "Harper Lee", "available": True},
    "1984": {"author": "George Orwell", "available": True},
    "Pride and prejudice": {"author": "Jane Austen", "available": True},
    "The great gatsby": {"author": "F. Scott Fitzgerald", "available": True},
    "Harry potter and the sorcerer's stone": {"author": "J.K. Rowling", "available": True},
    "Python crash course": {"author": "Eric Matthes", "available": True},
    "Sapiens": {"author": "Yuval Noah Harari", "available": True},
    "Atomic habits": {"author": "James Clear", "available": True}
}
```

```python
print("\n===== LIBRARY MANAGEMENT SYSTEM =====")
print("1. Add Book")
print("2. Borrow Book")
print("3. Return Book")
print("4. Search Book")
print("5. Display Books")
print("6. Exit")
print("=" * 37)
```

```python
choice = input("Enter your choice (1-6):")

# Option 1 : Add Book

if choice == "1":
    # Get the details of the book from user
    book_name = input("Please enter the name of the book:")
    author_name = input("Please enter the name of the author of the book:")
    # Add the book in the library dictionary
    if book_name.capitalize() not in library:
        library[book_name.capitalize()] = {"author": author_name, "available": True}
        print("Book added successfully")
    else:
        print("Book already available")
        
# Option 2 : Borrow a Book
elif choice == "2":

    # Get the name of the book from the user
    book_name = input("Please enter the name of the book:")

    # Check if the book is available in the library
    if book_name.capitalize() in library:
        if library[book_name.capitalize()]['available'] == True:
            # Set the avialability to False
            library[book_name.capitalize()]['available'] = False
            print("Book borrowed successfully")
        else:
            print('Book already Borrowed')
    else:
        print("Book doesn't exist in the libarary")

# Option 3 : Return a Book

elif choice == "3":
    # Get the name of book the user wants to return
    return_book = input("Please enter the name of the book:")
    # Check if the book exists in the linrary
    if return_book.capitalize() in library:
        # Check in the book is not available
        if library[return_book.capitalize()]['available'] == False:
            # Set the availability to True
            library[return_book.capitalize()]['available'] = True
            print("Book returned successfully")
        else:
            print("Book is already returned/not borrowed")
    else:
        print("Book doesn't exist in the library")

# Option 4 : Search for a Book

elif choice == "4":
    # Get the name of book the user wants to return
    search_book = input("Please enter the name of the book:")
    # Check if the book exists in the linrary
    if search_book.capitalize() in library:
        print(search_book , " available in library")
    else:
        print(search_book , " not available in library")

# Option 5 : Display all the books



elif choice == "5":
    available_books = 0
    not_available_books = 0

    for book_name in library.keys():
        print(book_name)
    for book_name in library.values():
        if book_name['available'] == True:
            available_books += 1
        else:
            not_available_books += 1

    print("Books available: ", available_books)
    print("Books not available: ", not_available_books)


# Option 6: Exit

elif choice == "6":
    print("Thank you for visiting")
```
