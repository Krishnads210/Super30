```python
emp1 = {"name": "Krishna", "salary_per_day": 1500, "days_present": 24, "late_days": 3}
emp2 = {"name": "Anil",    "salary_per_day": 1200, "days_present": 26, "late_days": 0}
emp3 = {"name": "Sravani", "salary_per_day": 1800, "days_present": 22, "late_days": 5}
emp4 = {"name": "Meera",   "salary_per_day": 2000, "days_present": 25, "late_days": 1}
emp5 = {"name": "Rahul",   "salary_per_day": 1350, "days_present": 20, "late_days": 4}

employees = [emp1,emp2,emp3,emp4,emp5]
```

```python
for i in employees:
    i['Gross_salary'] = i['salary_per_day']*30
    if i['late_days'] >= 0 and i['late_days'] <= 2:
        i['Late_deductions'] = 0
    elif i['late_days'] >= 3 and i['late_days'] <= 5:
        i['Late_deductions'] = 0.05*i['salary_per_day']*30
    elif i['late_days'] >= 6 and i['late_days'] <= 10:
        i['Late_deductions'] = 0.1*i['salary_per_day']*30
    elif i['late_days'] >= 10:
        i['Late_deductions'] = 0.2*i['salary_per_day']*30
    i['final_salary'] = i['Gross_salary']- i['Late_deductions']

```

```python
Salaries = []

for i in employees:
    Salaries.append(i['Gross_salary'])
    Highest_salary = sorted(Salaries)[-1]
    Lowest_salary = sorted(Salaries)[0]
    Total_company_payroll = sum(Salaries)

print(Salaries)
print(Highest_salary)
print(Lowest_salary)
print(Total_company_payroll)
```
